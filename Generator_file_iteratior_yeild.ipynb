```python
def fib(n):
    a = 1
    b = 1
    for i in range(n):
        yield a,i
        a,b = b ,a+b
```

```python
fib(9)
```

```python
tuple(fib(10))
```

```python
for i in fib(10):
    print(i)
```

```python
import re
def replace_word(file_name, old_word, new_word):
    with open(file_name, 'r+') as f:
        text = f.read()
        f.seek(0)
        f.write(re.sub(old_word, new_word, text))
        f.truncate()
replace_word('google.txt', 'Google', 'rakshanda')
```
