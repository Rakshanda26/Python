```python
# Overriding

class test:
    def __init__(self,a,b,c):
        self.a = a
        self.b = b
        self.c = c
        
    def __str__(self):
        return 'fdsfsd'  
    
    def class_fun(self):
        print('this is a print from class_fun')
        
class test1(test):
    def class_fun(self) :
        print('fsfsafas overriding method from test1')
        
```

```python
test1_obj = test1(1,2,3)
test_obj = test(4,5,6)
```

```python
test1_obj.class_fun()
```

```python
test_obj.class_fun()
```
